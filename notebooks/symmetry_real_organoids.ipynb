# Real Organoid Coarse Symmetry Scoring

This notebook mirrors `symmetry_synthetic_examples.ipynb`, but runs the extrinsic symmetry analysis on selected real organoid meshes from the dataset paths used in `test_curvature.ipynb`.

```python
import os
from pathlib import Path

import numpy as np

from organograph.mesh.OrganoidMesh import OrganoidMesh
from organograph.mesh.symmetry import (
    best_overall_symmetry,
    characteristic_size,
    describe_symmetry_score,
    laplace_beltrami_low_pass_vertices,
    pca_candidate_axes,
    reflection_matrix,
    rotation_matrix,
    run_multiscale_symmetry_analysis,
    sample_surface_points,
    score_transformation,
    surface_area_centroid,
    symmetry_results_to_records,
)
from organograph.plotting.meshes import plot_organoid_mesh
```

## Dataset Selection

The organoid IDs below are copied from the second cell of `test_curvature.ipynb`, including the commented-out examples. Edit `ORGANOIDS` to add/remove samples. The path builder follows the same directory convention as that notebook.

```python
NOTEBOOK_DIR = Path.cwd().resolve()

MESH_DATA_DIR = os.path.join(
    NOTEBOOK_DIR,
    "..",
    "..",
    "NicoleData",
    "20251201",
    "fractal_output",
)
TIMEPOINT = "day4p5"
ZARR_NAME = "251130R0.zarr"

ORGANOIDS = [
    {"well": "B03", "org_id": "144"},
    {"well": "B02", "org_id": "124"},
    {"well": "B04", "org_id": "4"},
    {"well": "B05", "org_id": "54"},
    {"well": "B02", "org_id": "115"},
    {"well": "B02", "org_id": "100"},
    {"well": "B02", "org_id": "31"},
    {"well": "B02", "org_id": "53"},
]


def organoid_key(record):
    return f"{record['well']}_{record['org_id']}"


def mesh_path_for(record):
    well = record["well"]
    org_id = record["org_id"]
    return (
        Path(MESH_DATA_DIR)
        / TIMEPOINT
        / ZARR_NAME
        / well[0]
        / well[1:]
        / "2_zillum_registered"
        / "meshes"
        / "nnorg_corrected_annotated_by_projection"
        / f"{org_id}.vtp"
    )


print("MESH_DATA_DIR =", MESH_DATA_DIR)
print("TIMEPOINT     =", TIMEPOINT)
for record in ORGANOIDS:
    path = mesh_path_for(record)
    status = "found" if path.exists() else "missing"
    print(f"{organoid_key(record):>8}  {status:>7}  {path}")
```

## Run Symmetry Analysis

Each mesh is loaded and normalized the same way as `test_curvature.ipynb`. The integer `l` values are Laplace-Beltrami reconstruction levels in the `OrganoidMesh` convention; each uses `l**2` spectral modes. `l=None` scores the normalized raw mesh.

The dataframe columns mean:

- `organoid_id`: sample key built from well and organoid ID.
- `l`: Laplace-Beltrami reconstruction level; `None` means the raw mesh was scored.
- `n_modes`: number of retained modes, equal to `l**2`; `None` for the raw mesh.
- `symmetry`: candidate symmetry family being scored: `reflection`, `C2` rotation, or `C3` rotation.
- `best_axis`: PCA candidate axis that gave the best score for that symmetry family at this smoothing level.
- `axis_x`, `axis_y`, `axis_z`: coordinates of the selected candidate axis or mirror-plane normal after centering.
- `trimmed_rms`: robust normalized mismatch; lower means stronger approximate coarse symmetry.
- `normalized_rms`: full RMS mismatch normalized by organoid size; lower is better but more outlier-sensitive.
- `median`: typical normalized mismatch.
- `matched_fraction`: sampled surface fraction within `close_threshold`; higher means more of the surface is explained by the symmetry.
- `characteristic_scale`: RMS surface radius used to make distances dimensionless.

```python
l_values = [None, 3, 5, 8]
max_l = max(l for l in l_values if l is not None)
n_samples = 8000

meshes_by_key = {}
results_by_key = {}
all_records = []
failed = []

for record in ORGANOIDS:
    key = organoid_key(record)
    mesh_path = mesh_path_for(record)
    if not mesh_path.exists():
        failed.append((key, "missing mesh file"))
        continue

    print(f"\n[{key}] loading {mesh_path}")
    try:
        mesh = OrganoidMesh(str(mesh_path))
        mesh.normalize_inplace()

        # Compute the largest requested spectral basis once; lower l levels reuse it.
        mesh.compute_spectral_coefficients(lmax=max_l)

        results = run_multiscale_symmetry_analysis(
            mesh,
            l_values=l_values,
            n_samples=n_samples,
            trim_fraction=0.95,
            close_threshold=0.05,
            random_seed=0,
        )
    except Exception as exc:
        failed.append((key, repr(exc)))
        print(f"[{key}] failed: {exc!r}")
        continue

    meshes_by_key[key] = mesh
    results_by_key[key] = results
    all_records.extend(symmetry_results_to_records(results, organoid_id=key))
    print(f"[{key}] {describe_symmetry_score(best_overall_symmetry(results))}")

try:
    import pandas as pd

    df = pd.DataFrame(all_records)
    with pd.option_context(
        "display.max_rows", None,
        "display.max_columns", None,
        "display.width", 0,
        "display.max_colwidth", None,
    ):
        display(df)
        if failed:
            display(pd.DataFrame(failed, columns=["organoid_id", "reason"]))
except ImportError:
    for row in all_records:
        print(row)
    if failed:
        print("failed:", failed)
```

## Best Symmetry Per Organoid

This is a compact summary of the best-scoring row per organoid. Treat it as a guide, not a hard label: check whether the winning score is clearly better than the alternatives and whether it remains stable across nearby `l` values.

```python
summary_rows = []
for key, results in results_by_key.items():
    best = best_overall_symmetry(results)
    summary_rows.append(best.to_record(organoid_id=key))
    print(f"{key}: {describe_symmetry_score(best)}")

try:
    import pandas as pd

    display(pd.DataFrame(summary_rows))
except ImportError:
    summary_rows
```

## Visualize All Tested Symmetry Transforms

Choose one processed organoid and one smoothing level. Columns are symmetry groups (`reflection`, `C2`, `C3`) and rows are candidate axes (`PCA1`, `PCA2`, `PCA3`). Each panel overlays the original low-pass surface in gray with the transformed surface for that specific candidate. The figure uses Plotly, so each panel can be rotated and zoomed interactively. For C3, the displayed transform is the 120-degree rotation; the score shown combines both 120 and 240 degrees.

```python
from scipy.spatial import cKDTree
from plotly.subplots import make_subplots
import plotly.graph_objects as go

display_organoid = next(iter(results_by_key)) if results_by_key else None
display_organoid = "B02_124"
display_l = 8

if display_organoid is None:
    raise RuntimeError("No processed organoids are available. Run the analysis cell first.")

mesh = meshes_by_key[display_organoid]
level_label = "raw mesh" if display_l is None else f"l={display_l}"

vertices = laplace_beltrami_low_pass_vertices(mesh, display_l)
faces = np.asarray(mesh.f, dtype=np.int64)
centroid = surface_area_centroid(vertices, faces)
centered_vertices = vertices - centroid[None, :]

viz_points = sample_surface_points(centered_vertices, faces, n_samples=8000, rng=17)
target_tree = cKDTree(viz_points)
scale = characteristic_size(viz_points)
candidate_axes = pca_candidate_axes(viz_points)
symmetry_groups = ["reflection", "C2", "C3"]


def matrices_for_symmetry(symmetry, axis):
    if symmetry == "reflection":
        return [reflection_matrix(axis)]
    if symmetry == "C2":
        return [rotation_matrix(axis, np.pi)]
    if symmetry == "C3":
        return [
            rotation_matrix(axis, 2.0 * np.pi / 3.0),
            rotation_matrix(axis, 4.0 * np.pi / 3.0),
        ]
    raise ValueError(symmetry)


def combined_candidate_score(matrices):
    scores = [
        score_transformation(
            viz_points,
            target_tree,
            matrix,
            scale,
            trim_fraction=0.95,
            close_threshold=0.05,
        )
        for matrix in matrices
    ]
    return {key: float(np.mean([score[key] for score in scores])) for key in scores[0]}


candidate_grid = []
subplot_titles = []
for axis_name, axis, _source in candidate_axes:
    row_entries = []
    for symmetry in symmetry_groups:
        matrices = matrices_for_symmetry(symmetry, axis)
        metrics = combined_candidate_score(matrices)
        row_entries.append((axis_name, axis, symmetry, matrices[0], metrics))
        trimmed = metrics["trimmed_rms"]
        matched = metrics["matched_fraction"]
        subplot_titles.append(
            f"{axis_name} | {symmetry}<br>trimmed RMS={trimmed:.3f}, matched={matched:.2f}"
        )
    candidate_grid.append(row_entries)

axis_limit = float(np.max(np.abs(centered_vertices)) * 1.08)
colors = {
    "reflection": "rgba(31, 119, 180, 0.45)",
    "C2": "rgba(255, 127, 14, 0.45)",
    "C3": "rgba(44, 160, 44, 0.45)",
}

fig = make_subplots(
    rows=len(candidate_axes),
    cols=len(symmetry_groups),
    specs=[[{"type": "scene"} for _ in symmetry_groups] for _ in candidate_axes],
    subplot_titles=subplot_titles,
    horizontal_spacing=0.01,
    vertical_spacing=0.04,
)

for row, row_entries in enumerate(candidate_grid, start=1):
    for col, (axis_name, axis, symmetry, matrix, metrics) in enumerate(row_entries, start=1):
        transformed_vertices = centered_vertices @ matrix.T
        axis = axis / np.linalg.norm(axis)
        axis_line = np.vstack([-axis_limit * axis, axis_limit * axis])
        show_legend = row == 1 and col == 1
        trimmed = metrics["trimmed_rms"]
        matched = metrics["matched_fraction"]

        fig.add_trace(
            go.Mesh3d(
                x=centered_vertices[:, 0],
                y=centered_vertices[:, 1],
                z=centered_vertices[:, 2],
                i=faces[:, 0],
                j=faces[:, 1],
                k=faces[:, 2],
                color="lightgray",
                opacity=0.24,
                name="original low-pass mesh",
                showlegend=show_legend,
                hoverinfo="skip",
            ),
            row=row,
            col=col,
        )
        fig.add_trace(
            go.Mesh3d(
                x=transformed_vertices[:, 0],
                y=transformed_vertices[:, 1],
                z=transformed_vertices[:, 2],
                i=faces[:, 0],
                j=faces[:, 1],
                k=faces[:, 2],
                color=colors[symmetry],
                opacity=0.45,
                name="transformed candidate",
                showlegend=show_legend,
                hovertemplate=(
                    f"{axis_name} | {symmetry}<br>"
                    f"trimmed RMS={trimmed:.3f}<br>"
                    f"matched={matched:.2f}<extra></extra>"
                ),
            ),
            row=row,
            col=col,
        )
        fig.add_trace(
            go.Scatter3d(
                x=axis_line[:, 0],
                y=axis_line[:, 1],
                z=axis_line[:, 2],
                mode="lines",
                line=dict(color="black", width=5),
                name="candidate axis / plane normal",
                showlegend=show_legend,
                hoverinfo="skip",
            ),
            row=row,
            col=col,
        )

scene_settings = dict(
    xaxis=dict(range=[-axis_limit, axis_limit], visible=False),
    yaxis=dict(range=[-axis_limit, axis_limit], visible=False),
    zaxis=dict(range=[-axis_limit, axis_limit], visible=False),
    aspectmode="cube",
    bgcolor="rgba(0,0,0,0)",
)
for row in range(1, len(candidate_axes) + 1):
    for col in range(1, len(symmetry_groups) + 1):
        fig.update_scenes(scene_settings, row=row, col=col)

fig.update_layout(
    title=f"{display_organoid}: all tested symmetry candidates on {level_label}",
    width=1320,
    height=1080,
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="left", x=0.0),
    margin=dict(l=0, r=0, b=0, t=110),
)
fig.show()
```

## Interpretation Notes

- Lower `trimmed_rms` means the transformed coarse surface lies closer to the original coarse surface.
- Higher `matched_fraction` means more of the surface is explained by the candidate symmetry within the relative tolerance.
- A strong coarse symmetry should remain reasonably stable across nearby `l` values.
- If the raw mesh has poor symmetry but the low-pass mesh improves, that suggests coarse biological symmetry with fine-scale epithelial/noise asymmetry.
- Reflection, C2, and C3 should not be forced into a hard label unless the winning score is clearly better than alternatives.
