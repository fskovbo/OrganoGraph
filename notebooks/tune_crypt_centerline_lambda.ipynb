# Tuning crypt centerline curvature regularization

This notebook scans the dimensionless curvature weight used by the endpoint-normal cubic Hermite centerline. Skeletonization is cached once for each organoid, and only primitive fitting is repeated. Overlap merging is disabled during the scan so every lambda is compared on the same crypt topology.

```python
from copy import deepcopy
from pathlib import Path

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML, display

from organograph.io_utils.dataset_config import load_mesh_dataset_config
from organograph.mesh.OrganoidMesh import OrganoidMesh
from organograph.mesh.geodesics import compute_geodesics_dijkstra
from organograph.plotting.skeletons import plot_mesh_with_skeleton_and_primitives
from organograph.skeleton import (
    SkeletonizationResult,
    definitive_filter_options,
    definitive_mesh_preparation,
    definitive_primitive_fit_config,
    definitive_skeletonization_config,
    fit_primitives_for_skeletonization_result,
    skeletonize_organoid,
)
from organograph.skeleton.primitive import crypt_terminal_paths
```

## 1) Experiment configuration

The three requested organoids are from dataset `20251201`. `LAMBDA_VALUES` controls the full scan; `DISPLAY_LAMBDAS` controls the heavier 3D comparison figures.

```python
NOTEBOOK_DIR = Path.cwd().resolve()
if NOTEBOOK_DIR.name != "notebooks" and (NOTEBOOK_DIR / "notebooks").exists():
    NOTEBOOK_DIR = NOTEBOOK_DIR / "notebooks"
PROJECT_ROOT = NOTEBOOK_DIR.parent

DATASET = "20251201"
DATASET_ROOT = PROJECT_ROOT.parent / "NicoleData" / DATASET
MESH_DATA_DIR = DATASET_ROOT / "fractal_output"
MESH_CONFIG_PATH = DATASET_ROOT / "mesh_config.json"
VOCAB_PATH = PROJECT_ROOT / "sim" / "vocab_with_meta.npz"

ORGANOID_SPECS = [
    {"timepoint": "day4p5", "well": "B03", "organoid_id": "43"},
    {"timepoint": "day4p5", "well": "B05", "organoid_id": "11"},
    {"timepoint": "day4p5", "well": "B02", "organoid_id": "43"},
]

LAMBDA_VALUES = np.array([0.0, 1e-4, 3e-4, 1e-3, 3e-3, 1e-2, 3e-2, 1e-1, 3e-1, 1.0])
DISPLAY_LAMBDAS = [0.0, 1e-3, 1e-2, 1e-1]
REFERENCE_LENGTH = None  # None: median attachment-to-tip chord across this test set
DISABLE_OVERLAP_MERGING_DURING_SCAN = True

mesh_cfg = load_mesh_dataset_config(str(MESH_CONFIG_PATH))
vocab = np.load(str(VOCAB_PATH), allow_pickle=True)
print("Dataset:", DATASET_ROOT)
print("Lambdas:", LAMBDA_VALUES)
```

```python
def mesh_path_from_spec(spec):
    timepoint = str(spec["timepoint"])
    well = str(spec["well"])
    return (
        MESH_DATA_DIR
        / timepoint
        / mesh_cfg["zarr_name_by_tp"][timepoint]
        / well[0]
        / well[1:]
        / mesh_cfg["round_by_tp"][timepoint]
        / "meshes"
        / mesh_cfg["meshname_by_tp"][timepoint]
        / f"{spec['organoid_id']}.vtp"
    )


organoid_records = []
for spec in ORGANOID_SPECS:
    record = dict(spec)
    record["label_uid"] = f"{spec['timepoint']}_{spec['well']}_{spec['organoid_id']}"
    record["mesh_path"] = str(mesh_path_from_spec(spec))
    if not Path(record["mesh_path"]).exists():
        raise FileNotFoundError(record["mesh_path"])
    organoid_records.append(record)

for record in organoid_records:
    print(record["label_uid"], record["mesh_path"])
```

## 2) Prepare meshes and cache skeletonization

```python
MESH_PREPARATION = definitive_mesh_preparation()
FILTER_OPTIONS = definitive_filter_options()


def _clamped_eigen_k(mesh, requested_k):
    return max(2, min(int(requested_k), int(np.asarray(mesh.v).shape[0]) - 2))


def _reset_spectral_state(mesh):
    mesh.laplacian = None
    mesh.mass_matrix = None
    mesh.eigvals = None
    mesh.eigvecs = None
    mesh.coeffs_v = None
    mesh.lmax = None


def _ensure_mesh_eigendecomposition(mesh, requested_k):
    k = _clamped_eigen_k(mesh, requested_k)
    if mesh.eigvals is None or mesh.eigvecs is None or mesh.eigvecs.shape[1] < k:
        _reset_spectral_state(mesh)
        mesh._eig_decomp(k=k)


def load_prepared_mesh(record):
    mesh = OrganoidMesh(record["mesh_path"])
    mesh.label_uid = record["label_uid"]
    if MESH_PREPARATION["normalize_mesh"]:
        mesh.normalize_inplace(scale=MESH_PREPARATION["normalize_scale"], center="mean")
    eigen_k = MESH_PREPARATION["eigen_k"]
    if MESH_PREPARATION["smooth_mesh"]:
        lmax = int(MESH_PREPARATION["smooth_lmax"])
        _ensure_mesh_eigendecomposition(mesh, max(eigen_k, lmax**2))
        mesh.compute_spectral_coefficients(lmax=lmax)
        mesh.v = np.asarray(mesh.reconstruct_from_coeffs(mesh.coeffs_v, lmax=lmax), dtype=float)
        _reset_spectral_state(mesh)
        _ensure_mesh_eigendecomposition(mesh, MESH_PREPARATION["smooth_eigen_k"] or eigen_k)
    else:
        _ensure_mesh_eigendecomposition(mesh, eigen_k)
    return mesh


def make_skeleton_config():
    config = definitive_skeletonization_config(filter_options=FILTER_OPTIONS)
    config.detection.barriers.attachment_strategy = "host_surface"
    return config


skeleton_results = {}
for record in organoid_records:
    display(HTML(f"<h3>Skeletonizing {record['label_uid']}</h3>"))
    mesh = load_prepared_mesh(record)
    skeleton_results[record["label_uid"]] = skeletonize_organoid(
        mesh,
        vocab,
        geodesic_fn=compute_geodesics_dijkstra,
        config=make_skeleton_config(),
        metadata={"record": dict(record), "label_uid": record["label_uid"]},
    )
```

## 3) Fix the reference length

A single reference length is used for every crypt and every lambda. This preserves the intended stronger physical penalty for concentrating the same bend into a short crypt.

```python
def unique_terminal_paths(graph):
    paths = {}
    for crypt_id in graph.crypt_ids():
        for path in crypt_terminal_paths(graph, crypt_id):
            paths[tuple(path)] = list(path)
    return list(paths.values())


chord_lengths = []
for skeleton in skeleton_results.values():
    for path in unique_terminal_paths(skeleton.graph):
        chord_lengths.append(
            float(np.linalg.norm(skeleton.graph.node(path[-1]).position - skeleton.graph.node(path[0]).position))
        )

if not chord_lengths:
    raise RuntimeError("No terminal crypt paths were found in the requested organoids")
CENTERLINE_REFERENCE_LENGTH = (
    float(np.median(chord_lengths)) if REFERENCE_LENGTH is None else float(REFERENCE_LENGTH)
)
print(f"Reference length: {CENTERLINE_REFERENCE_LENGTH:.4g}")
print(f"Crypt chord range: {min(chord_lengths):.4g} to {max(chord_lengths):.4g}")
```

## 4) Scan lambda

The fitting result records normalized centerline residual, physical bending energy, maximum curvature, curvature localization, and both tangent persistence lengths for every crypt.

```python
def clone_skeleton_result(source):
    return SkeletonizationResult(
        graph=deepcopy(source.graph),
        detections=deepcopy(source.detections),
        barriers=deepcopy(source.barriers),
        intermediates=deepcopy(source.intermediates),
        config=deepcopy(source.config),
        metadata=deepcopy(source.metadata),
        mesh=source.mesh,
        geodesic_fn=source.geodesic_fn,
    )


def primitive_config_for_lambda(value):
    config = definitive_primitive_fit_config()
    config.crypt_tube_kwargs["centerline_curvature_weight"] = float(value)
    config.crypt_tube_kwargs["centerline_reference_length"] = CENTERLINE_REFERENCE_LENGTH
    if DISABLE_OVERLAP_MERGING_DURING_SCAN:
        config.crypt_overlap.enabled = False
    return config


scan_results = {}
metric_rows = []
for lambda_value in LAMBDA_VALUES:
    display(HTML(f"<h3>lambda = {lambda_value:g}</h3>"))
    for label_uid, source in skeleton_results.items():
        result = fit_primitives_for_skeletonization_result(
            clone_skeleton_result(source),
            config=primitive_config_for_lambda(lambda_value),
        )
        scan_results[(label_uid, float(lambda_value))] = result
        for attachment_id, attachment in result.attachments["crypts"].items():
            metadata = attachment.metadata
            metric_rows.append(
                {
                    "label_uid": label_uid,
                    "attachment_id": str(attachment_id),
                    "lambda": float(lambda_value),
                    "normalized_rmse": float(metadata["centerline_fit_rmse"]) / CENTERLINE_REFERENCE_LENGTH,
                    "dimensionless_bending_energy": float(metadata["centerline_dimensionless_bending_energy"]),
                    "dimensionless_max_curvature": float(metadata["centerline_max_curvature"]) * CENTERLINE_REFERENCE_LENGTH,
                    "curvature_localization": float(metadata["centerline_curvature_localization"]),
                    "total_bend_angle_degrees": np.degrees(float(metadata["centerline_total_bend_angle"])),
                    "start_tangent_fraction": float(metadata["centerline_start_tangent_length"]) / CENTERLINE_REFERENCE_LENGTH,
                    "end_tangent_fraction": float(metadata["centerline_end_tangent_length"]) / CENTERLINE_REFERENCE_LENGTH,
                }
            )

print(f"Completed {len(scan_results)} organoid fits and {len(metric_rows)} crypt measurements")
```

## 5) Quantitative trade-off

Prefer the smallest lambda that removes the high-curvature tail and lowers curvature localization without causing a substantial increase in normalized contour RMSE.

```python
metrics = [
    ("normalized_rmse", "Normalized contour RMSE"),
    ("dimensionless_bending_energy", "Dimensionless bending energy"),
    ("dimensionless_max_curvature", "Dimensionless max curvature"),
    ("curvature_localization", "Curvature localization"),
]
fig = make_subplots(rows=2, cols=2, subplot_titles=[label for _, label in metrics])
for metric_index, (key, label) in enumerate(metrics):
    row, col = divmod(metric_index, 2)
    for lambda_value in LAMBDA_VALUES:
        values = [
            item[key] for item in metric_rows
            if np.isclose(item["lambda"], float(lambda_value)) and np.isfinite(item[key])
        ]
        fig.add_trace(
            go.Box(
                x=[f"{lambda_value:g}"] * len(values),
                y=values,
                name=f"{lambda_value:g}",
                boxpoints="all",
                jitter=0.25,
                marker_size=4,
                showlegend=False,
            ),
            row=row + 1, col=col + 1,
        )
    fig.update_xaxes(title_text="lambda", type="category", row=row + 1, col=col + 1)
    fig.update_yaxes(title_text=label, row=row + 1, col=col + 1)
fig.update_layout(height=760, width=1150, title="Centerline regularization scan", margin=dict(t=70))
display(fig)

median_points = []
for lambda_value in LAMBDA_VALUES:
    rows = [item for item in metric_rows if np.isclose(item["lambda"], float(lambda_value))]
    median_points.append(
        (
            float(lambda_value),
            float(np.median([item["normalized_rmse"] for item in rows])),
            float(np.median([item["dimensionless_bending_energy"] for item in rows])),
        )
    )
tradeoff = go.Figure(
    go.Scatter(
        x=[item[1] for item in median_points],
        y=[item[2] for item in median_points],
        text=[f"lambda={item[0]:g}" for item in median_points],
        mode="lines+markers+text",
        textposition="top center",
        hovertemplate="%{text}<br>median normalized RMSE=%{x:.4g}<br>median bending energy=%{y:.4g}<extra></extra>",
    )
)
tradeoff.update_layout(
    title="Median fit-versus-smoothness trade-off",
    xaxis_title="Median normalized contour RMSE",
    yaxis_title="Median dimensionless bending energy",
    width=850, height=560,
)
display(tradeoff)
```

## 6) Visual comparison

Change `DISPLAY_LAMBDAS` near the top to inspect additional scan values. These are independent figures so each organoid and lambda retains the same interaction and framing as the main skeleton tutorial.

```python
for record in organoid_records:
    label_uid = record["label_uid"]
    display(HTML(f"<h2>{label_uid}</h2>"))
    for lambda_value in DISPLAY_LAMBDAS:
        key = (label_uid, float(lambda_value))
        if key not in scan_results:
            print(f"lambda={lambda_value:g} was not included in LAMBDA_VALUES")
            continue
        result = scan_results[key]
        fig = plot_mesh_with_skeleton_and_primitives(
            result.mesh.v,
            result.mesh.f,
            result.graph,
            backend="plotly",
            mesh_alpha=0.12,
            primitive_alpha=0.32,
            show_node_labels=False,
        )
        fig.update_layout(
            title=f"{label_uid}: lambda={lambda_value:g}",
            showlegend=False,
        )
        display(fig)
```

## Selection rule

Use the visual comparisons to reject folds or biologically implausible paths. Among the remaining values, choose the smallest lambda near the knee of the trade-off plot, ideally before normalized RMSE rises by more than roughly 5-10% relative to lambda zero. Once selected, set `centerline_curvature_weight` and the same fixed `centerline_reference_length` in the definitive primitive profile.
